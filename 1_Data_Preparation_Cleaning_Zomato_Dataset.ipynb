```python
# ==========================================
# Time Standardization Function
# ==========================================

from datetime import datetime, timedelta
import pandas as pd
import numpy as np

def standardize_time(value):

    # Missing value
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    # ====================================
    # Case 1 : Excel Serial Time
    # ====================================
    try:
        if ":" not in value:
            excel_time = float(value)
            total_seconds = round(excel_time * 86400)
            hour = total_seconds // 3600
            minute = (total_seconds % 3600) // 60
            hour = hour % 24
            return f"{hour:02d}:{minute:02d}"
    except:
        pass

    # ====================================
    # Case 2 : HH:MM / HH:MM:SS
    # ====================================
    if ":" in value:
        parts = value.split(":")
        hour = int(parts[0])
        minute = int(parts[1])
        if hour == 24:
            hour = 0
        return f"{hour:02d}:{minute:02d}"
    return np.nan
```

```python
# ==========================================
# Standardize Time (Overwrite Original)
# ==========================================

df['Time_Orderd'] = df['Time_Orderd'].apply(standardize_time)
df['Time_Order_picked'] = df['Time_Order_picked'].apply(standardize_time)
```

```python
# ==========================================
# Haversine Formula
# ==========================================

from math import radians, sin, cos, sqrt, atan2

def haversine(lat1, lon1, lat2, lon2):

    R = 6371

    lat1 = radians(lat1)
    lon1 = radians(lon1)

    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        sin(dlat/2)**2
        +
        cos(lat1) *
        cos(lat2) *
        sin(dlon/2)**2
    )

    c = 2 * atan2(
        sqrt(a),
        sqrt(1-a)
    )

    return R * c
```

```python
# ==========================================
# Delivery Distance (KM)
# ==========================================

df['Delivery_Distance_km'] = df.apply(

    lambda row: haversine(

        row['Restaurant_latitude'],
        row['Restaurant_longitude'],

        row['Delivery_location_latitude'],
        row['Delivery_location_longitude']

    ),

    axis=1
)
```

### Coordinate Validation

During feature engineering, 156 records produced unrealistic delivery distances ranging from approximately 2,200 km to 6,900 km.

Further investigation showed that these records contained inconsistent geographic coordinates, where restaurant latitude was negative while delivery latitude remained positive, resulting in impossible delivery routes for a food delivery service.

Since the correct coordinates cannot be inferred, these records were treated as data quality issues and removed from further analysis.

```python
# ==========================================
# Remove Invalid Geographic Coordinates
# ==========================================

df = df[df['Restaurant_latitude'] >= 0].copy()

print(f"Remaining Records : {len(df):,}")
```

```python
# Hapus kolom lama
df.drop(columns=['Delivery_Distance_km'], inplace=True, errors='ignore')

# Hitung ulang
df['Delivery_Distance_km'] = df.apply(
    lambda row: haversine(
        row['Restaurant_latitude'],
        row['Restaurant_longitude'],
        row['Delivery_location_latitude'],
        row['Delivery_location_longitude']
    ),
    axis=1
)
```

```python
# ==========================================
# Time of Day
# ==========================================

def time_of_day(hour):

    if 5 <= hour < 12:
        return 'Morning'

    elif 12 <= hour < 17:
        return 'Afternoon'

    elif 17 <= hour < 21:
        return 'Evening'

    else:
        return 'Night'
```

```python
df['Time_of_Day'] = df['Order_Hour'].apply(time_of_day)
```
